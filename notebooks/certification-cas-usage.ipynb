## 1. Cadrage métier & problème

**Client** : Agence nationale de l'emploi

**Données** : 2500 echantillons provenant de l'agence nationale de l'emploi.

**Enjeux** : 
1. enjeu majeur éthique, reglementaire et métier 
    - avoir une prédiction fiable et limiter les erreurs de classification, surtout une classe 0 au lieu d'une classe 2 (Matrice de confusion)
    - usage de données personnelles et données sensibles, respect RGPD, non-discrimination, responsabilité juridique en cas d'erreur
2. enjeu sur la performance 
    - meilleur rapport performance / coût computationnel d'inférence (CPU/RAM)
    - latence acceptable
3. enjeu d'intégration dans un SI existant
    - intégration dans un SI existant et flux (API / DB usagers / AI)
    - souveraineté de l'infrastructure d'hébergement (on premise ou cloud souverain)
4. enjeu sur l'utilisabilité
    - interface graphique simple (saisie, prédiction et historique)

### 1.2 Énoncé du problème IA

Il s'agit d'un modèle supervisé de classification.
Prédire la cible X (3 classes) à partir de Y (8 données tablulaires et 1 donnée textuelle) avec une bonne précision et un taux d'erreur faible.
- Les 3 classes cibles sont : 
    - 0 : Rapide < 6 mois
    - 1 : Moyen entre 6 et 12 mois
    - 2 : Risque de longue durée > 12 mois).

### 1.5 Risques éthiques & réglementaires anticipés

**Cadre réglementaire applicable**

- **RGPD** : le traitement porte sur des données personnelles (identité, âge, commune, statut social, nationalité hors UE). Base légale retenue : **art. 6.1.e — mission d'intérêt public** (service public de l'emploi). À noter, car c'est une erreur fréquente : la **nationalité ne figure pas** dans la liste limitative des catégories particulières de l'**art. 9** (origine raciale ou ethnique, opinions politiques, convictions, appartenance syndicale, données génétiques, biométriques, de santé, vie ou orientation sexuelle). Le régime applicable est donc celui de l'art. 6, et non une dérogation de l'art. 9 — ce qui ne retire rien à la vigilance due, la nationalité pouvant constituer un **proxy d'origine**. S'y ajoutent la minimisation des données (`usager_id` ne doit pas être exploité comme feature ou être anonymisé), la durée de conservation à définir, et l'information des personnes concernées (art. 13-14).
- **AI Act (UE)** : un système qui influence l'accompagnement vers l'emploi d'un demandeur d'emploi relève probablement de l'**Annexe III — "emploi, gestion des travailleurs et accès au travail indépendant"**, donc potentiellement classé **à haut risque**. Cela implique : documentation technique, gestion des risques, supervision humaine effective (cf. §7.2 fallback), traçabilité (logs, cf. §0.5 et §8), et information des personnes concernées.
- **Droit sectoriel** : obligations de non-discrimination dans l'accès à l'emploi (Code du travail, Code pénal art. 225-1 sur les discriminations), et statut d'établissement public administratif de l'agence (transparence des décisions administratives, droit de recours de l'usager).

**Variables à risque (issues de `data_description.md`)**

| Variable | Risque identifié | Nature du risque |
|---|---|---|
| `usager_id` | Variable proxy si croisée avec une autre source (ré-identification) | RGPD — minimisation / ré-identification |
| `age` | Variable protégée (discrimination liée à l'âge) | Éthique + juridique (discrimination) |
| `niveau_diplome` | Corrélation possible avec l'origine sociale ; donnée manquante potentielle → biais d'imputation | Éthique (biais socio-économique) + qualité donnée |
| `anciennete_poste_ans` | Peut pénaliser les profils en reconversion / primo-demandeurs | Éthique (biais indirect) |
| `code_rome_vise` | Peut encoder des métiers genrés ou sectoriels (biais indirect de genre) | Éthique (biais indirect) |
| `code_insee_commune` | Variable protégée (localisation), Proxy de niveau de vie / zone géographique défavorisée (biais territorial) | RGPD + éthique (discrimination territoriale) |
| `est_allocataire` | Peut stigmatiser les profils indemnisés dans la décision | Éthique |
| `nationalite_hors_ue` | Proxy possible de l'origine ; critère protégé au titre de la non-discrimination (Code pénal art. 225-1) — **mais hors art. 9 RGPD** | Juridique et éthique — usage tranché par l'arbitrage `J0` (cf. encadré en fin de section) |
| `synthese_entretien` (texte libre) | Peut contenir des données sensibles non structurées (santé, situation familiale, origine perçue par le conseiller) et des biais subjectifs du conseiller | RGPD + éthique (biais humain encodé dans le texte) |

**Risques liés au modèle et à son usage**

- **Biais d'apprentissage** : le modèle peut apprendre des corrélations discriminatoires (ex. `nationalite_hors_ue` ou `code_insee_commune` corrélés à la classe prédite) même sans les utiliser explicitement (discrimination indirecte / proxy).
- **Erreur de classification asymétrique** : classer un usager en classe 0 (retour rapide) alors qu'il est en classe 2 (risque de longue durée) prive l'usager d'un accompagnement renforcé — impact humain et financier direct, à traiter en priorité dans la matrice de confusion (cf. §1.1, §6).
- **Automatisation excessive** : une décision d'orientation de l'accompagnement prise sans supervision humaine sur une donnée à fort enjeu social constitue un risque réglementaire (droit à une intervention humaine, art. 22 RGPD sur les décisions automatisées) et éthique.
- **Responsabilité juridique** : en cas d'erreur ayant causé un préjudice (mauvaise orientation, perte de chance), la responsabilité de l'agence peut être engagée — nécessité d'un plan de fallback documenté (§7.2) et d'une traçabilité des décisions (§0.5, §9).

---

**📌 Arbitrage `J0` — usage de `nationalite_hors_ue` comme feature (rendu après la phase 3)**

Cette section a d'abord posé, **avant tout entraînement**, une décision de précaution : ne pas utiliser la variable comme feature directe (cf. aussi §3.6, §3.8). La mesure *ex post* a renversé cette hypothèse, et la question a été portée devant la direction métier et la direction juridique / DPO. **Décision : l'usage est validé**, sous sept conditions.

*Pourquoi* : le §3.8 craignait que la variable serve à **pénaliser** les usagers hors UE. L'audit d'équité §7.2 établit l'inverse — recall classe 2 de **0.800 hors UE contre 0.529 UE**. La variable n'exclut pas ces usagers d'un dispositif, elle **améliore leur détection** et donc leur accès à l'accompagnement renforcé. L'exclure dégraderait la détection du public le plus exposé au nom de sa protection. S'y ajoutent quatre arguments : pas de traitement défavorable au sens de l'art. 225-1 (la sortie du modèle **déclenche une mesure favorable**, elle ne refuse rien) ; le dispositif relève de l'**action positive** admise en droit de l'Union (dir. 2000/43/CE art. 5) ; l'**AI Act art. 10(5)** autorise explicitement le traitement de données sensibles strictement nécessaire à la **détection et à la correction des biais** — or la variable est ici **à la fois** feature et axe obligatoire de l'audit §7.2, l'exclure priverait l'organisme de la capacité de mesurer le biais qu'il prétend éviter ; enfin, aucune décision individuelle automatisée au sens de l'**art. 22** (le modèle trie, il ne décide pas — l'arbitrage humain du §5.2.3 reste la condition de déploiement).

*Conditions attachées* : **C1** finalité strictement limitée à la priorisation vers un accompagnement renforcé (tout usage de contrôle, sanction, radiation ou refus est exclu) · **C2** interdiction d'exposition en supervision (aucun label Prometheus, aucun tableau Grafana, aucun log applicatif) · **C3** audit d'équité récurrent et outillé, avec marquage des effectifs sous le seuil de fiabilité de 30 · **C4** clause de retrait automatique si le recall classe 2 hors UE devient inférieur au recall UE · **C5** maintien de l'arbitrage humain §5.2.3 · **C6** AIPD (art. 35) et inscription au registre des traitements avant mise en service · **C7** information des personnes concernées (art. 13-14).

*Portée* : `J0` tranche **la seule réserve n° 2 du §7.3**. La réserve n° 1 — le filet de sécurité ne rattrape que 4 des 9 erreurs graves — reste **ouverte et bloquante** pour la mise en service. Le détail de la décision est consigné dans `decision.md`.

### 1.6 📝 Synthèse cadrage

Le problème posé par l'agence nationale de l'emploi est un problème de **classification supervisée multi-classe** : prédire, à partir de données hybrides (tabulaires + texte libre) sur 2500 usagers, le délai probable de retour à l'emploi en 3 classes (rapide, moyen, risque de longue durée). La tâche IA retenue est donc une classification à 3 classes combinés à un traitement du texte libre (`synthese_entretien`).

Le **critère de succès dominant** n'est pas l'accuracy globale mais la limitation des **erreurs asymétriques graves** — classer à tort un usager à risque de longue durée (classe 2) en retour rapide (classe 0) — car cette erreur prive l'usager d'un accompagnement renforcé et a un impact humain et financier direct ; le recall sur la classe 2 et le taux d'erreurs "classe 0 ↔ classe 2" priment donc sur l'accuracy brute.

Le **principal risque éthique** identifié est la présence de variables à risque ou proxy (`nationalite_hors_ue`, `code_insee_commune`, `age`, `synthese_entretien`) susceptibles d'introduire une discrimination directe ou indirecte dans la prédiction, dans un contexte où le système relève probablement d'un usage **à haut risque au sens de l'AI Act** (emploi/accès au travail) et impose une supervision humaine effective (cf. §7.2) plutôt qu'une automatisation intégrale de la décision.

#### 3.3.2.1 Démarche de traitement des commentaires

🎯 Objectif : transformer `synthese_entretien` en une variable exploitable par le modèle, après avoir vérifié **ce que ce texte est réellement**.

> 🔎 **Constat qui détermine toute la démarche (vérifié en §3.6 et re-contrôlé ci-dessous)** : `synthese_entretien` ne contient que **9 formulations uniques**, réutilisées telles quelles sur 2 419 des 2 500 lignes (81 valeurs manquantes). Le nettoyage ne fusionne aucune variante (`nunique` brut = `nunique` nettoyé = 9). Ce n'est donc **pas du texte libre** : c'est une **variable catégorielle à 9 modalités** (+ `texte_manquant`), connue et close.

**Conséquence** : aucun modèle NLP n'est nécessaire — ni TF-IDF, ni embeddings, ni zero-shot à l'inférence. Vectoriser 9 phrases en 300 termes dépenserait des dizaines de colonnes pour ré-encoder 10 états. Le texte est donc dérivé en une colonne `famille_thematique` puis encodé en **one-hot** par le pipeline tabulaire, comme n'importe quelle autre variable catégorielle (§4.2.1).

**Trace de la décision** : la classification **zero-shot CamemBERT** a bien été utilisée, mais **une seule fois**, hors pipeline, pour produire la correspondance template → famille. Son résultat est figé dans `data/referentiel_familles.csv` (9 lignes, avec score de confiance, modèle et date d'étiquetage) : l'étiquetage reste **traçable et auditable**, sans qu'aucun modèle NLP ne soit chargé à l'entraînement ou à l'inférence. Le détail de cette décision est consigné en §4.2.2.

**Étapes effectivement appliquées ci-dessous** :

1. **Préparation et protection des textes**
    * Remplacer les valeurs manquantes par une chaîne vide et conserver un indicateur `texte_manquant`.
    * Nettoyer uniquement les éléments nécessaires : espaces multiples et caractères parasites, sans supprimer le vocabulaire métier.
    * Rechercher et anonymiser les informations personnelles (téléphones, emails détectés par motifs).
    * Identifier les commentaires très courts ou peu informatifs (`commentaire_court`).
    * Quantifier les doublons exacts — c'est ce diagnostic qui a révélé les 9 templates.

2. **Dérivation de la famille thématique** (nouvelle étape, ex-« classification zero-shot »)
    * `pipeline_texte.assigner_famille(df)` : simple `map` sur le référentiel figé.
    * Les chaînes vides reçoivent la modalité `texte_manquant` ; **tout autre texte absent du référentiel lève une erreur**.
    * Un contrôle bloquant (`assert`) vérifie : **0 texte non mappé**, **9 familles** pour les textes non vides, **10 modalités** au total.

3. **NER — écarté**
    * Le NER n'est pas retenu : sur un corpus de 9 phrases connues, il n'apporterait ni familles thématiques supplémentaires, ni anonymisation utile (aucun nom, adresse ni organisme n'apparaît dans les templates).

* ⚠️ **Points de vigilance (maintenus)**
    * Les formulations peuvent encoder indirectement des variables sensibles (santé, situation familiale, origine, difficultés sociales). Ces éléments ne doivent pas devenir des critères de décision automatisée : l'audit par sous-groupe est fait en §3.6.2, et la famille thématique est reprise comme **axe d'équité** dans l'audit final (§7.2).
    * Vérifier l'absence de fuite de cible : aucun template ne mentionne directement le délai de retour à l'emploi ni une décision déjà prise par le conseiller.
    * La sensibilité du zero-shot à la formulation des catégories (cf. §4.2.2 : 7 templates sur 9 sous le seuil avec la taxonomie initiale) est **précisément la raison** pour laquelle aucun modèle NLP n'est conservé à l'inférence : un résultat non déterministe vis-à-vis du libellé des labels n'a pas sa place dans un pipeline de production.
    * Les textes manquants sont isolés dans une modalité dédiée (`texte_manquant`) et jamais affectés d'office à une famille.


````python
# Étape 1 : préparation et protection des textes de synthese_entretien

# Indicateur de texte manquant (avant remplacement)
df["texte_manquant"] = df["synthese_entretien"].isna()

# Remplacement des valeurs manquantes par une chaîne vide
df["synthese_entretien_prepare"] = df["synthese_entretien"].fillna("")

def nettoyer_texte(texte: str) -> str:
    """Nettoyage léger : espaces multiples et caractères parasites,
    sans supprimer le vocabulaire métier utile à l'analyse."""
    texte = re.sub(r"\s+", " ", texte)
    texte = re.sub(r"[^\w\sÀ-ÿ.,;:!?'-]", "", texte)
    return texte.strip()

df["synthese_entretien_prepare"] = df["synthese_entretien_prepare"].apply(nettoyer_texte)


# Détection des doublons exacts (hors chaînes vides)
mask_non_vide = df["synthese_entretien_prepare"] != ""
duplicated_mask = df.loc[mask_non_vide, "synthese_entretien_prepare"].duplicated(keep=False)
nb_doublons_texte = df.loc[mask_non_vide, "synthese_entretien_prepare"].duplicated().sum()
print(f"Nombre de doublons exacts de commentaires (hors vides) : {nb_doublons_texte}")
print(f"Nombre de lignes concernées par un doublon (au moins 1 autre occurrence) : {duplicated_mask.sum()} / {mask_non_vide.sum()}")

# Diagnostic : les textes dupliqués sont-ils de vrais doublons ou des templates réutilisés ?
print("\nTop des commentaires les plus répétés (diagnostic) :")
top_doublons = (
    df.loc[mask_non_vide, "synthese_entretien_prepare"]
    .value_counts()
    .head(10)
)
display(top_doublons)

print("\nNombre de valeurs uniques vs nombre total de commentaires non vides :")
print(f"Valeurs uniques : {df.loc[mask_non_vide, 'synthese_entretien_prepare'].nunique()}")
print(f"Total non vides : {mask_non_vide.sum()}")

# Détection d'informations personnelles potentielles (patterns simples à affiner avec le métier)
pattern_telephone = r"\b0[1-9](?:[\s.-]?\d{2}){4}\b"
pattern_email = r"[\w.+-]+@[\w-]+\.[a-zA-Z]{2,}"

df["contient_telephone"] = df["synthese_entretien_prepare"].str.contains(pattern_telephone, regex=True, na=False)
df["contient_email"] = df["synthese_entretien_prepare"].str.contains(pattern_email, regex=True, na=False)

def anonymiser_texte(texte: str) -> str:
    """Anonymisation basique : remplace téléphones et emails détectés.
    À enrichir avec un NER (§3.3.2.1 étape 3) pour noms, adresses, organismes."""
    texte = re.sub(pattern_telephone, "[TELEPHONE]", texte)
    texte = re.sub(pattern_email, "[EMAIL]", texte)
    return texte

df["synthese_entretien_prepare"] = df["synthese_entretien_prepare"].apply(anonymiser_texte)

# Identification des commentaires très courts ou peu informatifs (candidats "autre/ambigu")
SEUIL_LONGUEUR_MIN = 20  # caractères, seuil à valider avec le métier
df["commentaire_court"] = (
    (df["synthese_entretien_prepare"].str.len() < SEUIL_LONGUEUR_MIN)
    & (~df["texte_manquant"])
)

# Récapitulatif de la préparation
print(f"Textes manquants remplacés par une chaîne vide : {df['texte_manquant'].sum()}")
print(f"Doublons exacts détectés : {nb_doublons_texte}")
print(f"Commentaires contenant un pattern téléphone : {df['contient_telephone'].sum()}")
print(f"Commentaires contenant un pattern email : {df['contient_email'].sum()}")
print(f"Commentaires très courts (< {SEUIL_LONGUEUR_MIN} car.) : {df['commentaire_court'].sum()}")

df[["synthese_entretien", "synthese_entretien_prepare", "texte_manquant", "commentaire_court"]].head()

# Étape 2 : dérivation de la famille thématique (variable catégorielle, cf. §4.2.2)
# Simple `map` sur le référentiel figé `data/referentiel_familles.csv` : aucun modèle NLP
# n'est chargé ici (le zero-shot a servi une seule fois à produire ce référentiel).
df = pipeline_texte.assigner_famille(df)

# Contrôle bloquant : le contrat de `assigner_famille` doit être vérifié dans le notebook aussi.
textes_non_vides = df["synthese_entretien_prepare"].fillna("").str.strip() != ""
familles_observees = set(df.loc[textes_non_vides, "famille_thematique"].unique())
assert df["famille_thematique"].isna().sum() == 0, "Des lignes n'ont reçu aucune famille thématique."
assert pipeline_texte.FAMILLE_TEXTE_MANQUANT not in familles_observees, (
    "Un texte non vide est retombé sur la modalité 'texte_manquant' : référentiel incomplet."
)
assert len(familles_observees) == 9, (
    f"9 familles thématiques attendues pour les textes non vides, {len(familles_observees)} observées."
)
assert df["famille_thematique"].nunique() == 10, (
    "10 modalités attendues au total : 9 familles + 'texte_manquant'."
)

print(f"\nFamilles thématiques dérivées : {len(familles_observees)} familles + "
      f"'{pipeline_texte.FAMILLE_TEXTE_MANQUANT}' ({(~textes_non_vides).sum()} lignes sans texte)")
print(f"Textes non mappés : 0 (contrôle bloquant passé)")
display(df["famille_thematique"].value_counts())

````

=> Les données sont de bonne qualité : peu de manquants (81), aucun commentaire très court.

=> Aucune donnée sensible ne semble présente (téléphone, email).

=> Les commentaires sont normalisés : il n'y a que **9 valeurs uniques**, c'est-à-dire 9 types de commentaires, chacun présent plusieurs centaines de fois. C'est ce constat qui justifie de traiter `synthese_entretien` comme une **variable catégorielle** et non comme du texte libre (cf. §4.2.2).

=> La dérivation de `famille_thematique` est donc un simple `map` déterministe : les contrôles bloquants confirment **0 texte non mappé**, **9 familles** pour les textes non vides et **10 modalités** au total (`texte_manquant` inclus).

Deux questions à trancher :

* Le commentaire encode-t-il une variable protégée ? → croiser template × sous-groupe (est-ce que "barrière de la langue" est sur-représenté chez les nationalite_hors_ue == 1 ?).
* Le commentaire aggrave-t-il l'écart déjà mesuré sur la cible ? → comparer le taux de classe 2 par sous-groupe avec vs sans le signal texte dans le modèle (ça, c'est plutôt pour §6, après entraînement).

````python
# 3.6.x Audit des commentaires (synthese_entretien_prepare) par sous-groupe sensible
# Objectif : vérifier si certains templates de commentaires sont statistiquement
# associés à des variables protegees (nationalite_hors_ue, niveau_diplome, age_group),
# ce qui indiquerait un encodage indirect (proxy textuel).

sous_groupes_sensibles = ["nationalite_hors_ue", "niveau_diplome", "age_group"]

for variable_protegees in sous_groupes_sensibles:
    print(f"\n{'='*70}")
    print(f"Répartition des templates de commentaire selon {variable_protegees}")
    print(f"{'='*70}")

    table_croisee = pd.crosstab(
        df["synthese_entretien_prepare"],
        df[variable_protegees],
        normalize="index",
    ) * 100

    display(table_croisee.round(1))

    # Écart max entre sous-groupes, par template (diagnostic rapide)
    ecarts = (table_croisee.max(axis=1) - table_croisee.min(axis=1)).sort_values(ascending=False)
    print(f"\nTop 3 templates les plus discriminants pour {variable_protegees} :")
    print(ecarts.head(3).round(1))



    # 3.6.x (suite) Contrôle par la classe : l'association template x sous-groupe
# survit-elle à classe_retour_emploi constante ?
# Si oui -> vrai biais rédactionnel (encodage indirect réel).
# Si non (écarts qui s'effondrent) -> l'association passait entièrement par la cible.

for variable_protegees in sous_groupes_sensibles:
    print(f"\n{'='*70}")
    print(f"Écarts template x {variable_protegees}, À CLASSE CONSTANTE")
    print(f"{'='*70}")

    for classe in sorted(df["classe_retour_emploi"].unique()):
        sous_df = df[df["classe_retour_emploi"] == classe]

        # Ignorer les combinaisons avec trop peu d'effectif (peu fiable)
        if sous_df["synthese_entretien_prepare"].value_counts().min() < 5:
            print(f"  Classe {classe} : effectifs trop faibles par template, ignoré.")
            continue

        table_intra_classe = pd.crosstab(
            sous_df["synthese_entretien_prepare"],
            sous_df[variable_protegees],
            normalize="index",
        ) * 100

        ecarts_intra = (table_intra_classe.max(axis=1) - table_intra_classe.min(axis=1))
        print(f"  Classe {classe} — écart max moyen entre templates : {ecarts_intra.mean():.1f} points "
              f"(max {ecarts_intra.max():.1f})")

# 3.6.x (suite) Part de hors UE par template, à classe constante : le template sur-représente-t-il
# les usagers hors UE par rapport à la population de la MÊME classe (et non de tout le dataset) ?
resultats_part_hors_ue = []

for classe in sorted(df["classe_retour_emploi"].unique()):
    sous_df = df[df["classe_retour_emploi"] == classe]
    part_hors_ue_population = sous_df["nationalite_hors_ue"].mean()

    part_hors_ue_par_template = sous_df.groupby("synthese_entretien_prepare")["nationalite_hors_ue"].mean()
    effectif_par_template = sous_df.groupby("synthese_entretien_prepare").size()

    for template, part_template in part_hors_ue_par_template.items():
        resultats_part_hors_ue.append({
            "classe_retour_emploi": classe,
            "template": template,
            "effectif": effectif_par_template[template],
            "part_hors_ue_template": part_template * 100,
            "part_hors_ue_population_classe": part_hors_ue_population * 100,
            "ecart_points": (part_template - part_hors_ue_population) * 100,
        })

part_hors_ue_df = (
    pd.DataFrame(resultats_part_hors_ue)
    .sort_values(["classe_retour_emploi", "ecart_points"], ascending=[True, False])
    .reset_index(drop=True)
    .round(1)
)
print(f"\n{'='*70}")
print("Part de hors UE dans ce template vs part de hors UE dans la population, à classe constante :")
print(f"{'='*70}")
display(part_hors_ue_df)

# Effectifs associés aux plus gros écarts, pour juger de leur fiabilité statistique
SEUIL_EFFECTIF_FIABLE = 30  # en dessous, un écart est peu fiable (petit sous-groupe)
top_ecarts = part_hors_ue_df.reindex(part_hors_ue_df["ecart_points"].abs().sort_values(ascending=False).index).head(10)
print(f"\nTop 10 des écarts (en valeur absolue), avec effectif du template dans cette classe (seuil de fiabilité : {SEUIL_EFFECTIF_FIABLE}) :")
display(
    top_ecarts.assign(
        fiabilite=lambda d: d["effectif"].apply(lambda n: "fiable" if n >= SEUIL_EFFECTIF_FIABLE else "⚠️ petit effectif")
    )
)
````

=> À classe constante, l'écart entre la part de hors UE par template et la part de hors UE de la population de cette classe reste faible pour la plupart des templates, ce qui suggère que le lien observé plus haut (§3.6, association template × nationalité) passe essentiellement par la classe cible, et non par un biais rédactionnel propre au template. Quelques templates montrent toutefois un écart plus marqué (souvent sur les classes à faible effectif) : à interpréter avec prudence, ces sous-groupes restant de petite taille — cf. tableau des effectifs ci-dessus, qui distingue les écarts statistiquement fiables de ceux reposant sur un faible nombre d'observations.

**Pourquoi cette analyse est importante** : la corrélation brute mesurée en §3.6 entre les templates de commentaire et `nationalite_hors_ue` ne dit pas *où* se loge le risque de discrimination. Deux explications concurrentes sont possibles :
1. le conseiller rédige un commentaire différent selon la nationalité de l'usager, à situation professionnelle égale (biais rédactionnel direct) ;
2. le commentaire reflète simplement le profil professionnel de l'usager, qui est lui-même corrélé à la nationalité au niveau du dataset (biais indirect déjà capté par la classe cible).

> ⚠️ **Correction d'une prémisse fausse (relevé #11).** Ce paragraphe affirmait que « `nationalite_hors_ue` est déjà exclue des features (§4.1) » et en tirait l'innocuité du second cas. **C'est factuellement inexact** : la variable est une feature du scénario S1 (cf. `SCENARIO_FEATURES["s1"]`, §4.1, §6.1) et contribue effectivement à la prédiction (importance par permutation 0.030, §7.1). Le raisonnement est donc refait ci-dessous **sans** cette prémisse.

Le second cas reste **moins** problématique que le premier, mais pour une raison différente de celle invoquée initialement. Si le texte n'encode qu'un signal déjà porté par la classe cible, il n'ajoute pas de **canal supplémentaire** de discrimination : le risque n'est pas nul, il est simplement déjà couvert par l'audit d'équité §7.2, qui mesure le recall classe 2 **à la fois** par nationalité et par famille thématique. L'argument d'innocuité ne vient donc pas d'une exclusion de la variable — elle n'a pas lieu — mais de la **redondance** du signal et de sa mise sous surveillance explicite.

Le premier cas serait plus grave : il signifierait que le vocabulaire du conseiller varie selon l'origine de l'usager, indépendamment de sa situation réelle, et que le modèle pourrait apprendre ce biais via la famille thématique, qui est une feature du scénario S1 (cf. §7.1, §7.2). Ce risque-là se cumulerait à la contribution directe de `nationalite_hors_ue`, et non s'y substituerait.

**Ce que révèlent les chiffres** : en contrôlant par classe, l'écart de part hors UE entre un template et sa classe d'appartenance reste globalement modeste, ce qui penche en faveur de la seconde explication — le signal passe majoritairement par la classe, pas par une différence de traitement rédactionnel à situation égale. Les quelques écarts plus élevés se concentrent sur des templates à faible effectif (souligné par la colonne `fiabilite`) : leur ampleur peut provenir du bruit d'échantillonnage plutôt que d'un biais réel, et ne permet donc pas de conclure formellement. ````python

````python
TARGET_COLUMN = "classe_retour_emploi"
X = df.drop(columns=TARGET_COLUMN)
y = df[TARGET_COLUMN]

# 80 % des données servent à l'entraînement et 20 % sont réservées à l'évaluation finale.
# stratify=y conserve la proportion des trois classes dans X_train et X_test,
# notamment la classe 2 qui ne représente que 18,1 % du dataset.
# random_state=RANDOM_STATE rend le tirage reproductible à chaque exécution.
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    stratify=y,
    random_state=RANDOM_STATE,
)

print(f"Taille du jeu d'entraînement : {X_train.shape[0]} lignes ({X_train.shape[0] / len(df):.0%})")
print(f"Taille du jeu de test          : {X_test.shape[0]} lignes ({X_test.shape[0] / len(df):.0%})")

class_distribution = pd.DataFrame(
    {
        "dataset_complet": y.value_counts(normalize=True),
        "train": y_train.value_counts(normalize=True),
        "test": y_test.value_counts(normalize=True),
    }
).sort_index()

print("\nProportions des classes après le split :")
display((class_distribution * 100).round(1).rename_axis("classe_retour_emploi"))
````

##### 4.2.2.2 Étiquetage thématique des commentaires — **référentiel statique**

> ⚠️ **Décision structurante (phases 0 à 2)** : le texte `synthese_entretien_prepare` est traité comme une **variable catégorielle one-hot**. Il n'y a **plus aucun modèle NLP** dans le projet — ni zero-shot, ni TF-IDF.
>
> **Pourquoi.** Le texte ne contient que **9 formulations uniques** (§3.3.2.1, §3.6.2) couvrant 2 419 lignes sur 2 500 (81 manquants), et le nettoyage ne fusionne **aucune** variante (`nunique` brut = `nunique` nettoyé = 9). Le « texte libre » est en réalité une **variable catégorielle à 9 modalités**, connue et close. Deux conséquences :
> - faire tourner un transformeur de 68 M de paramètres à chaque exécution pour retrouver 9 étiquettes déjà connues est un coût sans contrepartie : ~2,5 Go de dépendances (`torch`, `transformers`, `sentencepiece`), un temps d'inférence non nul, et un résultat **non déterministe vis-à-vis de la formulation des labels** ;
> - vectoriser en **TF-IDF** ces 9 phrases est tout aussi inutile : avec `max_features=300` et `min_df=2`, le vocabulaire réellement construit ne comptait que **68 termes** — soit 68 colonnes pour encoder **10 états distincts**, sans aucune information supplémentaire.
>
> **Ce que le zero-shot a réellement produit.** Avec la taxonomie initiale (9 familles génériques : *mobilité géographique, formation, compétences numériques, expérience professionnelle, freins administratifs, santé, situation familiale, projet professionnel, autre*) et un seuil de 0,5, **7 templates sur 9** tombaient sous le seuil et repartaient en `a_valider` — soit **72,2 %** des lignes non étiquetées. La correspondance template → famille n'était donc **pas** exploitable.
>
> **Correction apportée.** En reformulant les 9 libellés pour qu'ils soient **mutuellement exclusifs et alignés sur le vocabulaire effectif des templates**, le même modèle atteint une correspondance **strictement 1-1** : 9 templates → 9 familles distinctes, **aucun doublon**, **aucun score sous le seuil** (score minimum **0,72**, maximum **0,93**). Cela confirme que l'échec initial venait de la **taxonomie**, pas du texte. Cette sensibilité à la formulation des catégories — anticipée dans les points de vigilance de §3.3.2.1 — est précisément la raison pour laquelle on ne conserve pas un modèle NLP à l'inférence.
>
> **Ce qui est conservé.** Le résultat de cet étiquetage, validé une fois pour toutes, est **figé dans un référentiel statique de 9 lignes** (`data/referentiel_familles.csv` : template, famille, score de confiance, modèle, date). #### 5.2.2 Courbes d'apprentissage & early stopping par scénario

🎯 Diagnostiquer, pour chaque scénario retenu ci-dessus, si le modèle souffre de sous-apprentissage (train et validation bas tous les deux) ou de sur-apprentissage (écart train/validation important), et si davantage de données d'entraînement pourrait encore aider.

💡 Seuls les 2 modèles à arbres (`RandomForestClassifier`, `HistGradientBoostingClassifier`) sont comparés ici (traits pleins = score train, pointillés = score validation CV), scénario par scénario. Score tracé : F1 macro, même métrique de référence qu'en §1.4/§5.3. `LogisticRegression` est exclu de cette comparaison : son solveur convexe n'a pas de notion d'itération qui « overfit », contrairement aux 2 modèles à arbres — cf. tableau ci-dessous.

⚠️ Sur `RandomForestClassifier`, le score train qui **diminue** quand la taille d'entraînement augmente (ex. S1, S4-age-dip) n'est pas une régression du modèle : un `RandomForestClassifier` non profondément régularisé mémorise quasi parfaitement un petit train set (score train ≈ 1.0), puis ce score redescend mécaniquement à mesure que le volume de données augmente car il devient plus difficile de mémoriser l'intégralité du bruit. C'est le signe attendu d'une réduction du sur-apprentissage — à lire conjointement avec le score de validation, qui lui **progresse** sur la plupart des scénarios (S1, S2, S4-age-dip-anc-dep, S4-age-dip). Un scénario resterait problématique seulement si le score de validation stagnait ou diminuait lui aussi avec plus de données.

Le même principe — « on arrête quand la validation cesse de progresser » — appliqué cette fois au **nombre d'itérations d'entraînement** (à taille de train fixe) plutôt qu'à la taille du train, se transpose différemment selon le modèle :

| Modèle | Équivalent d'un "epoch" | Early stopping appliqué ici |
|---|---|---|
| `HistGradientBoostingClassifier` | Un arbre de boosting ajouté à chaque itération | **Natif et activé** dans les courbes ci-dessous : `early_stopping=True` (+ `n_iter_no_change=3`, `validation_fraction=0.2`) arrête automatiquement le nombre d'itérations dès que la validation interne du modèle cesse de progresser, pour chaque taille de train testée. |
| `RandomForestClassifier` | Pas d'itération séquentielle par défaut (les arbres sont indépendants) | **Non applicable** : les arbres n'étant pas ajoutés séquentiellement, l'early stopping ne s'y transpose pas nativement ; son sur-apprentissage reste visible via l'écart train/validation de la courbe ci-dessous. |
| `LogisticRegression` | Itérations du solveur (`max_iter`) | **Non pertinent** : le solveur converge vers un optimum convexe, il n'« overfit » pas en ajoutant des itérations — exclu de la comparaison (son levier anti-surapprentissage est `C`, cf. §5.5). |


````python
# 5.2.2 Courbes d'apprentissage (train vs validation croisée) pour chaque scénario du benchmark ci-dessus.
# Modèles à arbres uniquement (RandomForestClassifier, HistGradientBoostingClassifier) : LogisticRegression
# est exclu ici, l'early stopping itératif ne le concernant pas (cf. tableau ci-dessus). Pour
# HistGradientBoostingClassifier, l'early stopping natif (early_stopping=True) est activé directement :
# à chaque taille de train testée, le nombre d'itérations s'arrête dès que sa validation interne stagne.
tailles_train_relatives = np.linspace(0.1, 1.0, 5)
scenarios_courbe_apprentissage = scenarios_tabulaires  # s1 et s3 inclus : un seul type de pipeline depuis la phase 2
PATIENCE = 3  # nombre d'itérations sans amélioration de la validation interne avant arrêt (HistGradientBoostingClassifier)

modeles_courbe_apprentissage = {
    "RandomForestClassifier": (RandomForestClassifier(class_weight="balanced", random_state=RANDOM_STATE), False),
    "HistGradientBoostingClassifier": (
        HistGradientBoostingClassifier(
            early_stopping=True, n_iter_no_change=PATIENCE, validation_fraction=0.2, random_state=RANDOM_STATE
        ),
        True,
    ),
}
couleurs_modeles = {
    "RandomForestClassifier": "tab:orange",
    "HistGradientBoostingClassifier": "tab:green",
}

fig, axes = plt.subplots(2, 3, figsize=(16, 9), sharey=True)
for ax, nom_scenario in zip(axes.flat, scenarios_courbe_apprentissage):
    for nom_modele, (modele_reference, besoin_dense) in modeles_courbe_apprentissage.items():
        pipeline_scenario = pipeline_tabulaire.build_scenario_pipeline(nom_scenario, clone(modele_reference), densify=besoin_dense)
        X_scenario = pipeline_tabulaire.prepare_tabular_features(X_train, nom_scenario)

        tailles_train_absolues, scores_train, scores_validation = learning_curve(
            pipeline_scenario, X_scenario, y_train,
            train_sizes=tailles_train_relatives, cv=cv, scoring="f1_macro", n_jobs=-1,
        )

        couleur = couleurs_modeles[nom_modele]
        ax.plot(tailles_train_absolues, scores_train.mean(axis=1), "o-", color=couleur, label=f"{nom_modele} (train)")
        ax.plot(tailles_train_absolues, scores_validation.mean(axis=1), "o--", color=couleur, label=f"{nom_modele} (validation)")

    ax.set_title(nom_scenario)
    ax.set_xlabel("Taille de l'échantillon d'entraînement")

for ax_inutilise in axes.flat[len(scenarios_courbe_apprentissage):]:
    ax_inutilise.set_visible(False)

axes.flat[0].set_ylabel("F1 macro")
handles, labels = axes.flat[0].get_legend_handles_labels()
fig.legend(handles, labels, loc="lower center", ncol=2, bbox_to_anchor=(0.5, -0.05))
plt.suptitle("Courbes d'apprentissage par scénario (RandomForestClassifier vs HistGradientBoostingClassifier avec early stopping natif)")
plt.tight_layout()
plt.show()
````

````python
# ⚠️ Comparaison sur X_sub_train (fit) / X_sub_val (évaluation) : X_test reste scellé et n'est utilisé
# qu'une seule fois, pour l'évaluation finale du modèle retenu.

# S1 inclut le texte sous forme de `famille_thematique` ; S4-all est la même liste de variables
# administratives SANS cette colonne. Un seul type de pipeline depuis la phase 2.
scenarios_finalistes = ["s1", "s4-all"]


def preparer_features_scenario(X, nom_scenario):
    """Features du scénario (la famille thématique y est une colonne comme une autre)."""
    return pipeline_tabulaire.prepare_tabular_features(X, nom_scenario)


def construire_pipeline_scenario(construire_modele, besoin_dense, nom_scenario):
    """Pipeline scikit-learn complet (préprocessing + modèle) du scénario."""
    return pipeline_tabulaire.build_scenario_pipeline(nom_scenario, construire_modele(), densify=besoin_dense)


lignes_comparaison_finalistes = []
lignes_perf_industrialisation = []

for nom_scenario in scenarios_finalistes:
    features_sub_train = preparer_features_scenario(X_sub_train, nom_scenario)
    features_sub_val = preparer_features_scenario(X_sub_val, nom_scenario)

    for nom_base, variantes in MODELES_FINALISTES.items():
        for construire_modele, besoin_dense, description in variantes:
            nom_modele = f"{nom_base} ({description})"
            pipeline_perf = construire_pipeline_scenario(construire_modele, besoin_dense, nom_scenario)

            # Étape 1 — train : fit chronométré sur le sous-train
            temps_fit_s = train_module.train(pipeline_perf, features_sub_train, y_sub_train)
            # Étape 2 — save : sérialisation du pipeline + taille disque
            taille_mo, chemin_modele = train_module.save(pipeline_perf, nom_scenario, nom_modele)
            # Étape 3 — latence : temps d'inférence unitaire p50/p95
            latences_ms = train_module.latence_performance(pipeline_perf, features_sub_val.iloc[[0]])
            latence_p50_ms = round(float(np.percentile(latences_ms, 50)), 3)
            latence_p95_ms = round(float(np.percentile(latences_ms, 95)), 3)

            # Étape 4 — métriques §1.4 sur la sous-validation (sert à choisir le modèle final ci-dessous)
            resultat_comparaison = metrics_module.evaluate_model(pipeline_perf, features_sub_val, y_sub_val)
            # Récupération des coûts métier déjà calculés en §5.5 (5-fold CV) pour ce couple (scénario, modèle)
            resultat_cv_precedent = resultats_benchmark_modeles.get((nom_scenario, nom_modele), {})
            for cle_cout in metrics_module.COST_METRIC_LABELS:
                if cle_cout in resultat_cv_precedent:
                    resultat_comparaison[cle_cout] = resultat_cv_precedent[cle_cout]
            # Caractéristiques d'industrialisation (§8/§9.1) ajoutées à la comparaison des finalistes.
            resultat_comparaison["taille_mo"] = round(taille_mo, 3)
            resultat_comparaison["temps_fit_s"] = round(temps_fit_s, 3)
            resultat_comparaison["latence_p50_ms"] = latence_p50_ms
            resultat_comparaison["latence_p95_ms"] = latence_p95_ms
            lignes_comparaison_finalistes.append(
                metrics_module.metrics_to_row(nom_scenario, resultat_comparaison, modele=nom_modele)
            )

            lignes_perf_industrialisation.append({
                "scenario": nom_scenario,
                "modele": nom_modele,
                "taille_mo": round(taille_mo, 3),
                "temps_fit_s": round(temps_fit_s, 3),
                "latence_p50_ms": latence_p50_ms,
                "latence_p95_ms": latence_p95_ms,
                "chemin_modele": str(chemin_modele),
            })
````

### 6.3 📝 Synthèse arbitrages

Le compromis qui a tranché le choix : retenir la configuration qui **maximise la détection des cas à risque** (recall classe 2 = 0.681 et F1 classe 2 = 0.653 en sous-validation, les meilleures valeurs des finalistes) **tout en maintenant le taux d'erreur grave sous 10 %** (9,7 % en sous-validation, 10,0 % sur le test set), plutôt que de viser le minimum absolu d'erreurs graves au prix de la détection.

Quatre arbitrages structurent cette décision :

1. **Représentation du texte : one-hot plutôt que TF-IDF ou zero-shot** — c'est l'arbitrage le plus structurant du projet, et il ne coûte rien. Le tableau avant/après de §5.2.1 bis le démontre : sur S3, les métriques sont **strictement identiques** entre 68 colonnes TF-IDF et 10 modalités one-hot ; sur S1, la version one-hot est même légèrement **meilleure** (+0.027 de F1 macro et −1,6 point d'erreur grave pour `RandomForestClassifier`). On supprime donc 2,5 Go de dépendances, une inférence de transformeur et 58 colonnes de bruit **sans perte de performance**, tout en gagnant en explicabilité.
2. **Texte vs sécurité pure** — ajouter la famille thématique fait passer le recall classe 2 de 0.514 (S4-all) à 0.681 (S1) en sous-validation, au prix d'un taux d'erreur grave qui remonte de 4,2 % à 9,7 %. Ce surcoût de risque est jugé acceptable parce qu'il reste sous 10 % **et** qu'il est encadré par une revue manuelle ciblée ; l'inverse (rater la moitié des usagers à risque) ne l'est pas.
3. **Texte seul écarté** — S3 atteint un recall équivalent (0.660) mais avec un taux d'erreur grave de 19,1 %, soit le double de S1. La famille thématique n'apporte sa valeur qu'**en complément** des variables administratives, pas en remplacement — et ce malgré le fait qu'elle soit, prise isolément, la variable la plus importante du modèle (§7.1).
4. **Éthique vs performance** — S2, le scénario sans aucune variable sensible ni proxy, est le plus défendable éthiquement mais reste nettement dominé (18,5 % d'erreurs graves contre 9,7 %). Nous conservons donc l'ensemble des variables de S1, y compris `nationalite_hors_ue`. C'est l'arbitrage le plus lourd du projet : il est assumé au titre de la performance, mais il **transfère la charge de la maîtrise du biais sur l'audit d'équité (§7.2) et sur la revue manuelle**, et non sur la sélection des features.

Entre les finalistes sur S1, `RandomForestClassifier(n_estimators=300)` l'emporte sur `HistGradientBoostingClassifier(class_weight=balanced, max_depth=10)` : il détecte 5 usagers à risque de plus sur 90 (recall 0.681 contre 0.625), gagne 0.020 de F1 macro et 10 410 € de coût métier, au prix de 1,4 point d'erreur grave supplémentaire (9,7 % contre 8,3 %, les deux restant sous la barre) et d'un modèle 34× plus lourd (27,5 Mo contre 0,81 Mo). **À noter — cet arbitrage a changé de sens** : dans la version TF-IDF, c'était `HistGradientBoostingClassifier` qui était retenu ; le passage au one-hot a nettement profité à la forêt aléatoire, qui exploite mieux une variable catégorielle propre qu'une matrice creuse de termes.

**Ce choix reste conditionné** à la mise en place effective de la revue manuelle sur les ~26 % de dossiers en alerte (§7.2) et à la validation de l'audit d'équité par sous-groupe — lequel inclut désormais un axe supplémentaire, la **famille thématique** elle-même, rendu possible par la nouvelle représentation du texte.

### 7.1 Importance des variables

Le modèle retenu (§5.6/§6.1) est `RandomForestClassifier(n_estimators=300, class_weight="balanced")` sur le scénario **S1** : 7 variables administratives (`age`, `nationalite_hors_ue`, `niveau_diplome`, `anciennete_poste_ans`, `code_rome_vise`, `est_allocataire`, `departement`) **+ `famille_thematique`**, la synthèse d'entretien réduite à une variable catégorielle à 10 modalités.

L'importance est mesurée **par permutation** : on mélange les valeurs d'une variable et on observe la perte de F1 macro qui en résulte. Plus la perte est forte, plus le modèle s'appuyait sur cette variable. On préfère cette méthode aux `feature_importances_` natives de la forêt, qui sont biaisées en faveur des variables à forte cardinalité (ici `code_rome_vise` et `departement`) et qui portent sur les colonnes **après** encodage, pas sur les variables métier.

> 🔁 **Ce que la phase 2 change ici, et c'est décisif.** Dans la version TF-IDF, la synthèse d'entretien occupait 68 colonnes de termes : il fallait les permuter « en bloc » pour obtenir un chiffre lisible, et l'on ne pouvait pas dire *quels* termes pesaient ni les auditer par sous-groupe. Désormais, **le texte EST une colonne d'entrée unique** : la permutation de `famille_thematique` mesure directement et sans artifice l'apport du commentaire du conseiller, à la même granularité que `age` ou `niveau_diplome`. L'explicabilité du modèle n'est plus dégradée par le traitement du texte.


````python
# 7.1 Importance des variables du modèle final (scénario S1, modèle retenu en §5.6/§6.1)
# L'importance est mesurée par permutation sur le pipeline complet, au niveau des colonnes d'entrée
# métier. Depuis la phase 2, la synthèse d'entretien EST une colonne d'entrée unique et interprétable
# (`famille_thematique`, 10 modalités) : il n'y a plus de bloc TF-IDF à agréger artificiellement,
# la permutation de cette colonne mesure donc directement l'apport du texte.
importance_permutation = permutation_importance(
    pipeline_final,
    features_test_final,
    y_test,
    scoring="f1_macro",
    n_repeats=10,
    random_state=RANDOM_STATE,
    n_jobs=-1,
)

importance_df = (
    pd.DataFrame(
        {
            "variable": features_test_final.columns,
            "perte_f1_macro": importance_permutation.importances_mean.round(4),
            "ecart_type": importance_permutation.importances_std.round(4),
        }
    )
    .sort_values("perte_f1_macro", ascending=False)
    .reset_index(drop=True)
)

print("Importance par permutation sur le test set (perte de F1 macro quand la variable est mélangée) :")
display(importance_df)

plt.figure(figsize=(8, 4.5))
plt.barh(
    importance_df["variable"][::-1],
    importance_df["perte_f1_macro"][::-1],
    xerr=importance_df["ecart_type"][::-1],
    color=["#d62728" if v in {"nationalite_hors_ue", "age", "niveau_diplome", "departement"}
           else "#2ca02c" if v == "famille_thematique" else "#1f77b4"
           for v in importance_df["variable"][::-1]],
)
plt.axvline(0, color="grey", linewidth=0.8)
plt.xlabel("Perte de F1 macro quand la variable est mélangée")
# Le libellé du modèle est DÉRIVÉ du pipeline effectivement évalué, jamais recopié :
# c'est ainsi que ce titre avait fini par annoncer un HistGradientBoostingClassifier
# alors que `pipeline_final` est un RandomForest (relevé #12).
_modele_final = pipeline_final.named_steps["model"]
_params_final = _modele_final.get_params()
_descripteur_final = "{}({})".format(
    type(_modele_final).__name__,
    ", ".join(
        f"{cle}={_params_final[cle]!r}"
        for cle in ("n_estimators", "class_weight", "random_state")
        if cle in _params_final
    ),
)
plt.title(
    f"Importance par permutation — {_descripteur_final}\n"
    "scénario S1 (tabulaire + famille thématique), test set\n"
    "Rouge = variable sensible ou proxy (§1.5/§3.6) — Vert = synthèse d'entretien (famille_thematique)"
)
plt.tight_layout()
plt.show()
````

Conséquence opérationnelle : les seuils A et B doivent être **recalibrés avec le métier sur un volume de production**, en acceptant d'augmenter la part de dossiers revus si l'on veut couvrir davantage d'erreurs graves.

⚠️ Les ~74 % de dossiers en `"auto"` conservent le risque d'erreur résiduel du modèle brut : c'est l'hypothèse explicite du calcul de coût métier de §5.2.3 (les dossiers revus sont supposés corrigés sans erreur, les autres non). Cette hypothèse doit être revalidée en production via §9.2.

**Audit d'équité — résultats, et ils sont préoccupants.** L'audit croise le recall classe 2 avec quatre axes de sous-groupe, dont **un nouveau rendu possible par la phase 2** : la famille thématique elle-même, qui est désormais une feature explicite du modèle.

* **Nationalité hors UE** (variable sensible **et** feature du modèle) : recall classe 2 de **0.800** pour les usagers hors UE (16/20) contre **0.529** pour les autres (37/70), soit **27 points d'écart**. Le modèle détecte donc *mieux* le risque chez les usagers hors UE — ce qui est favorable en termes d'accompagnement, mais constitue bel et bien un **traitement différencié fondé sur une variable protégée**, et doit être arbitré par le métier et le juridique, pas par le data scientist.
* **Niveau de diplôme** : recall de **0.829** pour les usagers sans diplôme (29/35) contre **0.385 à 0.469** pour les titulaires d'un Bac, Bac+2 ou Bac+5 — jusqu'à **44 points d'écart**.
* **Tranche d'âge** : recall de **0.756** pour les 51-60 ans (31/41) contre **0.333** pour les 41-50 ans et **0.438** pour les 21-30 ans.
* **Famille thématique** (nouvel axe) : le recall varie de **0.000** (« réactualisation des compétences sur les outils numériques », 0/7) à **0.786** (« garde d'enfants et absence de moyen de transport », 11/14). Les familles associées à des freins lourds (« freins périphériques et illettrisme numérique » 0.739, « perte de confiance et barrière de la langue » 0.762) sont bien détectées ; celles qui décrivent des situations intermédiaires le sont beaucoup moins. **C'est une information directement actionnable pour le métier** : elle indique aux conseillers quels types de situation le modèle repère mal — un diagnostic strictement impossible à produire avec la représentation TF-IDF précédente.

**Lecture d'ensemble.** Ces écarts sont **cohérents entre eux** : sans diplôme, 51-60 ans, freins périphériques lourds et nationalité hors UE désignent largement la même population, celle où la classe 2 est la plus fréquente et la plus « lisible » dans les données. Le modèle est donc performant là où le signal est net, et faible sur les profils intermédiaires — un comportement attendu, mais qui produit mécaniquement une **inégalité de service** : un cadre de 45 ans en difficulté a nettement moins de chances d'être repéré qu'un usager sans diplôme de 55 ans.

⚠️ **Réserve statistique impérative** : avec 90 cas de classe 2 au total sur le test set, plusieurs sous-groupes comptent moins de 10 observations (seuil de fiabilité §3.6 = 30). Aucun de ces écarts ne peut être considéré comme établi. Ils constituent un **signal d'alerte à reconfirmer sur volume de production**, pas une conclusion d'équité.

Le code ci-dessous quantifie ce filet sur le test set : combien de dossiers sont signalés, quelle part des erreurs graves est effectivement rattrapée, et sur quels profils portent les erreurs résiduelles.


````python
# 7.2 Efficacité du filet de sécurité et audit d'équité (test set, déjà ouvert en §5.6.2 — aucune
# nouvelle évaluation ici, on réutilise le pipeline final et les mêmes règles §5.2.3 que le benchmark).
predictions_test = pipeline_final.predict(features_test_final)
probabilites_test = pipeline_final.predict_proba(features_test_final)

# Toutes les colonnes d'entrée du modèle sont directement exploitables comme axes d'audit
# (y compris `famille_thematique`, désormais une variable catégorielle interprétable).
analyse_erreurs = features_test_final.copy()
analyse_erreurs["classe_reelle"] = y_test.to_numpy()
analyse_erreurs["classe_predite"] = predictions_test
analyse_erreurs["proba_classe_2"] = probabilites_test[:, 2]
analyse_erreurs["confiance_max"] = probabilites_test.max(axis=1)
analyse_erreurs["erreur_grave_2_vers_0"] = (analyse_erreurs["classe_reelle"] == 2) & (
    analyse_erreurs["classe_predite"] == 0
)
# Règles §5.2.3 (seuil A = 0.40, seuil B = 0.20), appliquées par la même fonction que le benchmark.
analyse_erreurs["a_valider"] = metrics_module.appliquer_regles_validation_manuelle(
    predictions_test, probabilites_test, seuil_A, seuil_B
)
analyse_erreurs["regle_b_seule"] = (
    (analyse_erreurs["classe_predite"] == 0)
    & (analyse_erreurs["proba_classe_2"] >= seuil_B)
    & (analyse_erreurs["proba_classe_2"] < seuil_A)
)

part_a_valider_test = analyse_erreurs["a_valider"].mean()
print(
    f"Dossiers signalés pour revue manuelle : {analyse_erreurs['a_valider'].sum()} / {len(analyse_erreurs)} "
    f"({part_a_valider_test:.1%}) — coût de revue {analyse_erreurs['a_valider'].sum() * COUT_REVUE_MANUELLE} € "
    f"à {COUT_REVUE_MANUELLE} €/dossier"
)
print(f"  dont {analyse_erreurs['regle_b_seule'].sum()} déclenchés par la seule règle B (priorité haute, erreur grave potentielle)")

# Combien d'erreurs graves le filet rattrape-t-il réellement ?
erreurs_graves = analyse_erreurs["erreur_grave_2_vers_0"]
rattrapees = erreurs_graves & analyse_erreurs["a_valider"]
print(
    f"\nErreurs graves 2→0 : {erreurs_graves.sum()} au total, dont {rattrapees.sum()} signalées pour revue "
    f"({rattrapees.sum() / max(erreurs_graves.sum(), 1):.0%}) et {(erreurs_graves & ~analyse_erreurs['a_valider']).sum()} "
    "non rattrapées (risque résiduel après filet)"
)

print("\nConfiance max du modèle — prédictions correctes vs erreurs graves :")
display(
    pd.DataFrame(
        {
            "correctes": analyse_erreurs.loc[
                analyse_erreurs["classe_reelle"] == analyse_erreurs["classe_predite"], "confiance_max"
            ].describe(),
            "erreurs_graves_2_vers_0": analyse_erreurs.loc[erreurs_graves, "confiance_max"].describe(),
        }
    ).round(3)
)


def recall_classe_2_par_groupe(df: pd.DataFrame, colonne: str) -> pd.DataFrame:
    """Recall de la classe 2 par modalité, avec l'effectif de classe 2 pour juger de la fiabilité."""
    est_classe_2 = df["classe_reelle"] == 2
    detectes = est_classe_2 & (df["classe_predite"] == 2)
    tableau = pd.DataFrame(
        {
            "effectif_classe_2": est_classe_2.groupby(df[colonne], observed=True).sum(),
            "detectes": detectes.groupby(df[colonne], observed=True).sum(),
        }
    )
    tableau["recall_classe_2"] = (tableau["detectes"] / tableau["effectif_classe_2"]).round(3)
    return tableau


# Audit d'équité (prérequis de mise en production, §5.7/§6.3) : nationalite_hors_ue est ici une
# FEATURE du modèle S1 (§6.1), pas seulement une variable de contrôle — l'écart de recall est donc
# directement imputable au modèle.
analyse_erreurs["tranche_age"] = pd.cut(analyse_erreurs["age"], bins=age_bins, labels=age_labels, right=False)

for colonne, libelle in [
    ("nationalite_hors_ue", "nationalité hors UE (feature du modèle, variable sensible)"),
    ("niveau_diplome", "niveau de diplôme (feature du modèle, proxy §1.5)"),
    ("tranche_age", "tranche d'âge (dérivée de age, feature du modèle, proxy §1.5)"),
    ("famille_thematique", "famille thématique de la synthèse d'entretien (feature du modèle, §4.2.2)"),
]:
    print(f"\nRecall classe 2 selon {libelle} :")
    display(recall_classe_2_par_groupe(analyse_erreurs, colonne))

print(f"\nRecall classe 2 selon departement (top 10 par effectif, feature du modèle) :")
top_departements = analyse_erreurs["departement"].value_counts().head(10).index
display(
    recall_classe_2_par_groupe(
        analyse_erreurs[analyse_erreurs["departement"].isin(top_departements)], "departement"
    ).sort_values("effectif_classe_2", ascending=False)
)

print(
    f"\n⚠️ Les effectifs de classe 2 par sous-groupe sont faibles (90 cas au total sur le test set, "
    f"seuil de fiabilité §3.6 = {SEUIL_EFFECTIF_FIABLE}) : ces écarts sont indicatifs et doivent être "
    "reconfirmés sur un volume de production avant toute conclusion d'équité."
)
````

### 7.3 Message au client (langage métier)

**Ce que fait l'outil.** Il analyse le dossier administratif de l'usager (âge, diplôme, ancienneté, métier visé, statut d'allocataire, département, nationalité hors UE) **et** la synthèse de l'entretien rédigée par le conseiller, pour estimer un risque de chômage de longue durée. Il classe chaque situation en trois niveaux : retour rapide à l'emploi, retour intermédiaire, ou risque de longue durée. C'est la combinaison dossier + entretien qui donne les meilleurs résultats : sans la synthèse d'entretien, l'outil ne repère plus qu'un cas à risque sur deux.

**Un mot sur le traitement de l'entretien.** L'outil ne « lit » pas le commentaire du conseiller au sens d'une intelligence artificielle générative. Nous avons constaté que les conseillers utilisent en pratique **9 formulations types**, toujours les mêmes. L'outil se contente donc de reconnaître laquelle a été employée — une correspondance figée, vérifiable ligne à ligne, sans modèle de langage. C'est à la fois plus sobre, plus rapide, entièrement reproductible, et cela nous permet de mesurer précisément l'effet de chaque type de commentaire sur la décision. C'est d'ailleurs **l'information la plus déterminante du dossier** : elle pèse près de deux fois plus que l'âge dans la décision de l'outil.

**Ce qu'il sait faire, et ce qu'il rate.** Sur 90 usagers réellement à risque dans notre jeu de test, l'outil en repère correctement **53, soit 59 %**. Il se trompe gravement — classer un usager à risque en « retour rapide » — dans **10 % des cas**, soit 9 usagers sur 90. C'est le meilleur équilibre détection/sécurité que nous ayons obtenu, mais il reste **au-dessus de notre objectif de sécurité initial de 5 %**, et en deçà de notre objectif de détection (80 %).

**Comment on encadre ce risque.** L'outil ne décide pas : il trie. Environ **un dossier sur quatre (26 %)** est automatiquement signalé pour vérification par un conseiller — soit parce que le risque estimé est élevé, soit parce que l'outil annonce un retour rapide alors qu'un doute subsiste. Les trois dossiers sur quatre restants suivent la recommandation sans vérification systématique. Sur notre jeu de test, cette vérification ciblée représente 2 560 € de temps conseiller pour 500 dossiers (environ 5 € par dossier traité), à comparer au coût d'un accompagnement engagé trop tard. Techniquement, l'outil répond en moins de 50 millisecondes : il s'intègre sans difficulté dans les applicatifs existants.

**⚠️ Deux réserves à lever avant mise en service.**

1. **Le filet de sécurité ne suffit pas encore.** Sur les 9 erreurs graves observées, **4 seulement sont signalées pour vérification** : 5 passent à travers. L'outil se trompe parfois avec une confiance élevée, ce qu'aucun réglage de seuil ne rattrape totalement. Les règles d'alerte doivent être recalibrées avec vos équipes sur un volume réel, en acceptant probablement d'augmenter la part de dossiers vérifiés.
2. **Les écarts de détection entre profils — ✅ arbitrés depuis (arbitrage `J0`).** L'outil repère mieux le risque chez certains publics que chez d'autres : 83 % chez les usagers sans diplôme contre 39 à 47 % chez les diplômés ; 80 % chez les usagers de nationalité hors UE contre 53 % chez les autres ; 76 % chez les 51-60 ans contre 33 % chez les 41-50 ans. Ces écarts reposent sur des effectifs faibles et doivent être reconfirmés, mais ils signifient qu'**à situation comparable, tous les usagers n'ont pas la même chance d'être repérés**. Cet arbitrage relevait de vous et de votre service juridique : **il a été rendu**. Nous recommandons, sous réserve de validation par vos directions métier et juridique l'usage de la nationalité comme critère de repérage, en constatant qu'il **joue en faveur** des usagers hors UE (80 % de détection contre 53 %) et qu'il sert à leur ouvrir un accompagnement, jamais à le restreindre. Sept conditions y sont attachées, dont l'interdiction de tout autre usage, un audit d'équité récurrent, et le retrait automatique de la variable si l'écart venait à s'inverser (cf. `decision.md`).

> ⚠️ **Cette levée ne vaut pas feu vert.** Seule la réserve n° 2 est levée. **La réserve n° 1 reste ouverte et bloquante** : tant que le filet de sécurité ne rattrape que 4 des 9 erreurs graves, la mise en production ne doit pas être engagée.

### 7.4 📝 Synthèse interprétation

**Variables-clés.** L'importance par permutation (§7.1) est mesurée au niveau des colonnes métier. Depuis la refonte du traitement du texte, la synthèse d'entretien **est** une variable unique (`famille_thematique`, 10 modalités) et non plus un bloc de 68 colonnes TF-IDF : son apport est donc mesuré directement, sans agrégation artificielle. Résultat : elle arrive **première, et de loin** (perte de F1 macro 0.170), devant `age` (0.095) et `niveau_diplome` (0.089), et près de 6× devant `nationalite_hors_ue` (0.030). C'est cohérent avec le benchmark (§5.3/§6.1), où l'ajout du texte fait passer le recall classe 2 de 0.514 (S4-all) à 0.681 (S1) en sous-validation. À l'inverse, `est_allocataire` (−0.001) et `departement` (−0.006) sont **non contributifs** : leur retrait ne coûterait rien et supprimerait un proxy territorial identifié en §1.5 — une piste d'amélioration éthique gratuite.

**Profils mal prédits.** L'audit de §7.2 produit le recall classe 2 par nationalité UE/hors UE, tranche d'âge, diplôme, département **et famille thématique** (axe rendu possible par la nouvelle représentation du texte). Les écarts sont importants et cohérents entre eux : le modèle détecte bien les profils où le risque est « lisible » (sans diplôme 0.829, 51-60 ans 0.756, freins périphériques lourds 0.739-0.762) et mal les profils intermédiaires (41-50 ans 0.333, Bac+2 0.385, famille « réactualisation des compétences numériques » 0.000 sur 7 cas). Cette analyse reste **indicative** : avec 90 cas de classe 2 sur le test set, la plupart des sous-groupes passent sous le seuil de fiabilité retenu en §3.6 (30 observations). La conclusion d'équité doit être reconfirmée sur un volume de production — c'est le point ouvert identifié en §5.7 et reporté en condition de déploiement en §6.3.

**Vigilance opérationnelle.**
1. **Le taux d'erreur grave (10,0 %) est au double de la cible de 5 %** : la revue manuelle ciblée sur ~26 % des dossiers (règles A et B de §5.2.3) est la condition qui rend ce niveau acceptable. Supprimer ou réduire cette revue invalide l'arbitrage de §6.3.
2. **Le filet ne rattrape que 44 % des erreurs graves** (4 sur 9) : le risque résiduel est explicite et quantifié. Les erreurs graves surviennent avec une confiance moyenne de 0.591 contre 0.729 pour les prédictions correctes — un écart trop faible pour être pleinement discriminant. À suivre en production (§9.4).
3. **`nationalite_hors_ue` est une feature du modèle**, pas seulement une variable de contrôle (§6.1), et elle contribue effectivement à la prédiction (importance 0.030). L'écart de recall entre usagers UE (0.529) et hors UE (0.800) est donc directement imputable au modèle, ce qui rend l'audit d'équité bloquant.
4. **La famille thématique peut encoder des proxys indirects** — mais, contrairement à la version TF-IDF, ce risque est désormais **mesurable** : §3.6.2 l'audite à classe constante, et §7.2 mesure le recall par famille. Le point n'est plus « non quantifié », il est quantifié et sous surveillance.
5. **Aucune explication individuelle n'est disponible à ce stade** : l'importance par permutation donne une lecture globale du modèle, pas la justification d'une prédiction précise. Pour motiver une décision auprès d'un usager, une méthode d'attribution locale (SHAP) serait nécessaire — non réalisée ici.
6. **Aucune dépendance NLP ne subsiste** : plus de `torch`, `transformers` ni `TfidfVectorizer` (§4.2.2, §5.0). L'inférence est entièrement déterministe et le service est allégé de ~2,5 Go de dépendances, ce qui réduit d'autant la surface de maintenance, de sécurité et l'empreinte d'exécution.

**Implémenté** : serveur MLflow Tracking adossé au `docker-compose` (backend SQLite, artefacts sur volume `mlruns/`), exposé sur le port 5000. La façade `src/tracking.py` pousse vers lui les métadonnées **déjà produites** par `src/train.py::save_metadata` et l'arbitrage de `scripts/retrain.py`.

⚠️ **Le tracking est strictement optionnel, par conception.** Si `MLFLOW_TRACKING_URI` n'est pas définie, si le paquet `mlflow` est absent ou si le serveur est injoignable, toutes les fonctions deviennent des no-op et **aucune exception ne remonte à l'appelant**. Perdre la trace d'un run est ennuyeux ; perdre un entraînement de 40 minutes parce qu'un serveur de métadonnées était en panne serait inacceptable. Aucun service applicatif ne dépend de MLflow, et le paquet est absent des images de service : le tracking est une activité **hors ligne**, un conteneur d'inférence n'a rien à écrire dans un serveur d'expériences.

**Pourquoi MLflow plutôt que la solution précédente.** Le projet traçait ses expériences par des **livrables markdown générés** (`baseline.md`, `benchmark.md`, `comparaison_finalistes.md`, `evaluation_finale.md` via `src/metrics.py::write_benchmark_markdown`) et par un **journal de décisions** append-only (`decisions_log.jsonl`). Ce dispositif est reproductible, diffable en revue de code et lisible par un jury — il **reste la source de vérité des livrables**, et `decisions_log.jsonl` reste le journal qui fait foi. Il lui manquait trois choses, et trois seulement :

1. le **rattachement artefact ↔ run** autrement que par un champ `git_commit` recopié ;
2. la **comparaison multi-runs** sans régénérer un markdown ;
3. un **cycle de vie de modèle** (`Staging` → `Production`) sur lequel brancher `scripts/promotion.py`.

💡 Le choix se fait par rapport à l'alternative crédible, pas dans l'absolu : DVC versionne bien les données mais suit mal les métriques de modèle ; W&B est SaaS, donc écarté d'emblée pour un traitement fondé sur l'art. 6.1.e RGPD portant sur des demandeurs d'emploi. MLflow est auto-hébergeable — c'est le critère décisif ici. Le backend SQLite (plutôt qu'un simple `file:`) est imposé par le *Model Registry*, qui exige un backend relationnel.

⚠️ **Un rejet de promotion est tracé au même titre qu'une acceptation.** Ne journaliser que les promotions reviendrait à ne garder que les runs qui arrangent. Le rejet est d'ailleurs l'issue la plus instructive : une série de rejets pour « gain insuffisant » dit que le flux de feedback n'apporte plus d'information, un rejet pour « régression critique » dit exactement l'inverse.

### 8.3 Interface utilisateur

**Dossier** : `services/frontend/` — formulaire conseiller HTML/JS statique servi par **nginx**, qui fait aussi **reverse-proxy** de `/api/` vers `backend:8001`.

**Pourquoi pas Streamlit ni Gradio.** Ces outils sont excellents pour une démo exploratoire, mais ils embarquent un runtime Python, un état serveur par session et un modèle de rendu qu'on ne maîtrise pas. Or l'interface visée ici est un **formulaire de saisie à 8 champs** destiné à un poste de travail de conseiller. Trois arguments ont tranché :

1. **Surface d'attaque et empreinte** : une image `nginx:1.27-alpine` pèse quelques Mo et n'exécute pas de code applicatif. Un conteneur Streamlit embarque tout l'écosystème Python côté public.
2. **Même origine, donc pas de CORS navigateur** : le reverse-proxy `/api/` fait que le front et l'API partagent l'origine. Le `ALLOWED_ORIGINS` du BFF reste une défense en profondeur, pas la seule barrière.
3. **Testabilité du contrat** : `services/frontend/tests/test_frontend.py` vérifie que les champs du formulaire correspondent **exactement** au schéma Pydantic et que les options des listes déroulantes correspondent aux **modalités réellement apprises par l'encodeur**. Ce test est impossible à écrire proprement sur une UI générée dynamiquement — et c'est lui qui empêche la dérive silencieuse la plus fréquente en production : un formulaire qui propose une modalité que le modèle n'a jamais vue.

**Information des personnes concernées (condition `C7` de `J0`)** : le champ `nationalite_hors_ue` est isolé dans un `<fieldset>` distinct portant une mention explicite de son caractère sensible, de sa finalité (priorisation vers un accompagnement renforcé) et de son interdiction d'usage hors de ce traitement (condition `C1`).

**La quality gate n'est pas une répétition des tests.** Les trois barrières répondent à trois questions distinctes :

| Barrière | Question | Nature |
|---|---|---|
| `tests/test_modele_servi.py` | *les fichiers se contredisent-ils ?* | concordance documentaire, ne charge aucune donnée |
| `scripts/promotion.py` | *ce candidat mérite-t-il de remplacer la prod ?* | arbitrage entre **deux** modèles, dans la boucle de rétroaction |
| `scripts/quality_gate.py` | *l'artefact qu'on met en image tient-il encore ses promesses ?* | **rejoue** le golden run depuis le `.joblib` servi |

⚠️ La gate est le seul contrôle du dépôt capable de détecter un `reference_baseline.json` **périmé**. Les suites comparent ce fichier à des constantes et à `THRESHOLDS` : elles restent vertes si le fichier et les constantes dérivent ensemble, ou si l'artefact est remplacé sans régénérer la baseline. Seule la gate confronte un fichier de métriques à une **prédiction réellement calculée** — sur les 350 lignes de référence, avec une tolérance de `1e-4`, trois ordres de grandeur sous celle de `promotion.py` (le golden run est déterministe, il n'y a pas de bruit d'échantillonnage à absorber).

Les 7 contrôles : absence de clé de neutralisation ou de second jeu de métriques (`G1.1`), version servie alignée sur le golden run (`G1.2`), contrat d'entrée à 8 champs (`G1.3`), `build_pipeline()` reproduisant la configuration servie (`G2.1`), concordance avec la métadonnée du notebook (`G3.1`), golden run rejoué (`G4.1`), planchers de promotion franchis par le modèle servi lui-même (`G4.2`).

**Le job `smoke-e2e` couvre ce qu'aucune suite pytest ne peut couvrir.** Les tests unitaires éprouvent chaque service avec ses voisins bouchonnés ; trois pannes ne sont observables qu'à l'assemblage — une dépendance absente de l'image, un `depends_on` mal ordonné, une URL de reverse-proxy erronée — et toutes se manifesteraient en production. L'appel de bout en bout passe donc par `:8088/api/score`, c'est-à-dire par nginx, et non par le backend en direct : appeler le backend laisserait la configuration nginx non testée. Le job vérifie aussi **positivement** qu'un payload à 7 champs renvoie `422` (sans quoi un schéma redevenu permissif passerait inaperçu) et que `nationalite_hors_ue` n'apparaît **ni dans les logs applicatifs, ni dans `/metrics`** — c'est la contrepartie `C2` de `J0`, vérifiée en exécution et pas seulement en revue.

**Le job `deploy` ne reconstruit rien** : il promeut par `docker buildx imagetools create` les images que `build-and-push` a produites et que `smoke-e2e` a validées. Reconstruire à ce stade déploierait un binaire que personne n'a testé. Il est derrière un `environment: production` — la protection de branche ne suffit pas, c'est l'environnement qui porte la validation manuelle et les secrets de la cible.

`fail-fast: false` sur les tests est délibéré : quand une modification du schéma casse le contrat, on veut voir **quelles** suites tombent (model seul ? model + frontend ?), pas seulement la première.

**Pourquoi FastAPI plutôt que Flask.** Trois raisons, dans l'ordre d'importance pour ce projet : (1) **Pydantic v2 est intégré au framework**, donc la validation des 8 champs est le contrat, pas une couche ajoutée — Flask aurait exigé marshmallow et une discipline de cohérence manuelle ; (2) **l'OpenAPI est généré depuis les schémas**, donc la documentation ne peut pas diverger du code, ce qui compte quand trois services partagent un contrat ; (3) **ASGI natif**, ce qui permet d'attendre le chargement de l'artefact dans un `lifespan` et de répondre `503` tant qu'il n'est pas prêt — comportement que le `healthcheck` du compose exploite directement.


### 8.6 📝 Synthèse industrialisation

**Pile retenue** : scikit-learn (pipeline sérialisé joblib) · FastAPI + Pydantic v2 (3 services) · nginx (UI + reverse-proxy) · SQLite (feedback) · Docker Compose (7 conteneurs) · Prometheus + Grafana · MLflow (tracking, hors ligne) · GitHub Actions (tests → quality gate → smoke e2e → build GHCR → deploy sur tag).

**Les trois décisions d'architecture qui structurent tout le reste :**

1. **Le pipeline complet est sérialisé, pas seulement l'estimateur.** Il n'existe donc aucun code de préprocessing dans le périmètre en ligne, et aucun risque d'écart entraînement/service (*training-serving skew*) — la cause n°1 de dégradation silencieuse en production.
2. **Trois services plutôt qu'un monolithe.** Le service d'inférence est ignorant du métier et remplaçable ; le BFF absorbe les pannes amont et porte les compteurs métier ; la collecte de vérité terrain est isolée avec son propre stockage. Le coût est une duplication assumée du schéma Pydantic, verrouillée par des tests de contrat.
3. **La boucle de rétroaction ne se referme pas toute seule.** Réentraînement, décision de promotion et promotion effective sont trois étapes distinctes, dont la deuxième est une fonction pure testée sur un golden set figé. Sur un système relevant de l'Annexe III de l'AI Act, l'automatisation complète du cycle serait un défaut de conception, pas une maturité.

**Ce que l'industrialisation garantit** : le modèle servi est **exactement** celui évalué en §5.6.2 / §6.1 / §7.1 / §7.2 — même configuration, mêmes 8 features, mêmes métriques, vérifié à la fois par test automatisé et par rejeu effectif du golden run avant toute mise en image. Le `request_id` propagé de bout en bout rend l'audit d'équité du §7.2 **rejouable sur données de production**, ce qui est la condition `C3` de `J0`.

### 9.2 Boucle de rétroaction

Après observation de la situation réelle, un conseiller associe au `request_id` du dossier la classe constatée (`0`, `1` ou `2`) via `POST /feedback`. Le service vérifie que l'identifiant est connu, enregistre le retour dans SQLite et expose le nombre de retours total et non encore consommés via `/feedback/count`. Un rejeu du même label ne crée pas de doublon ; un label contradictoire renvoie `409` et requiert un arbitrage humain.

Le `request_id` relie feedback et dossier scoré. `scripts/feedback_store.py` joint les annotations aux features du journal ; elles servent à l'audit hors ligne et à l'entraînement candidat. Les jeux d'exemple sont simulés : leur vérité terrain provient du dataset connu et ne représente pas un retour réel de conseiller.

**Limite du prototype à lever avant une exploitation réelle :** `/predict` renvoie un `request_id`, mais ne l'écrit pas encore dans `data/prod_scored.csv`, qui est actuellement un artefact statique de démonstration. Le raccordement durable du scoring en ligne au journal et le canal opérationnel de saisie conseiller restent à réaliser. Le fichier scoré et la base de feedback peuvent contenir des données personnelles : leurs accès et durées de conservation doivent être encadrés.

### 9.3 Plan de réentraînement

**Cadence opératoire retenue : contrôle hebdomadaire manuel** du nombre de nouveaux feedbacks. Lorsque le seuil de 100 annotations non consommées est atteint, l'opérateur lance `python scripts/retrain.py --min-feedback 100`. Aucun cron ni ordonnanceur n'est configuré dans le dépôt ; le déclenchement hebdomadaire est une procédure à organiser, pas une automatisation existante. Sous le seuil, le script ne fait rien.

Le script entraîne un candidat en combinant le jeu d'entraînement d'origine et les nouveaux dossiers annotés, puis vérifie le contrat de prédiction (probabilités dans [0, 1], classes attendues). Le candidat et le modèle de production sont évalués sur le **même jeu de référence figé**. `scripts/promotion.py` n'accepte le candidat que s'il respecte les planchers, ne régresse pas de plus de 0,01 sur les métriques critiques (recall classe 2 et erreur grave 2→0) et gagne au moins 0,01 sur une métrique suivie.

Chaque décision, acceptation comme rejet, est ajoutée à `decisions_log.jsonl` ; MLflow reçoit aussi l'arbitrage s'il est disponible. Un candidat rejeté ne remplace pas le modèle servi et ses feedbacks restent disponibles. En cas d'acceptation, un artefact versionné distinct est produit ; le déploiement effectif reste une étape séparée et soumise à revue humaine. Il n'y a ni apprentissage en ligne ni promotion automatique.

### 9.4 Robustesse en exploitation

🎯 Surveiller les **conditions de défaillance** une fois en prod, en complément des stratégies de fallback définies en §7.2 (côté conception).

💡 La conception du fallback (où placer le seuil, qui escalader) vit en §7.2. Ici on décrit les **mesures disponibles** et les décisions à prendre après investigation ; aucun déclenchement automatique n'est configuré :

| Axe | Mesure disponible dans le dépôt | Interprétation et suite |
|---|---|---|
| Disponibilité et erreurs | Santé des cibles, latence, débit et réponses 4xx/5xx dans Grafana | Détecter visuellement une panne ou une dégradation ; aucune notification automatique n'est en place. |
| Drift d'entrée | Distribution des `famille_thematique` sur 15 minutes | Comparer aux fréquences de référence. Un écart invite à examiner les pratiques de saisie et les dossiers ; aucun score PSI ni seuil d'alerte n'est implémenté. |
| Drift de sortie | Répartition des classes prédites et histogramme de la probabilité de la classe prédite | Un changement de distribution est un signal d'investigation, pas une preuve de baisse de qualité. Sans labels récents, F1 et erreurs ne sont pas mesurables. |
| Qualité et équité après annotation | Rejeu hors ligne sur les feedbacks via `scripts/audit_equite.py` | Mesurer le recall classe 2 par sous-groupe, identifier les effectifs sous 30 et contrôler la clause C4. Si le recall hors UE devient inférieur au recall UE avec effectifs suffisants, le script retourne le code 2 ; le retrait effectif de la variable et le traitement de l'alerte ne sont pas automatisés. |

Le service ne calcule pas de score OOD, de taux d'abstention en exploitation, de Brier score ni de courbe de calibration. La distribution des probabilités exposée par Prometheus ne permet pas, à elle seule, de conclure à une bonne calibration. Ces contrôles ne sont donc pas présentés comme acquis.

_Note: "score OOD" : (Out-Of-Distribution) estime à quel point une nouvelle entrée est différente des données sur lesquelles le modèle a été entraîné ; "Brier score" :  mesure la qualité des probabilités prédites par un modèle de classification. Il compare les probabilités annoncées aux résultats réellement observés, une fois les étiquettes connues.._

⚠️ La chaîne conception → exploitation doit rester explicite. §7.2 définit le filet de sécurité, §9.4 le surveille. Casser cette chaîne (fallback conçu mais jamais mesuré) = attendu C8 N3 non atteint.

### 9.5 📝 Synthèse amélioration continue

**Rituel hebdomadaire proposé :** consulter Grafana (disponibilité, erreurs, latence, classes prédites et familles thématiques), vérifier le volume d'annotations et lancer le batch candidat si 100 nouveaux feedbacks sont disponibles. À réception de labels, rejouer l'audit de qualité et d'équité hors ligne avant toute décision de promotion. Cette revue n'est pas planifiée automatiquement dans le dépôt.

En cas d'indisponibilité ou de hausse des erreurs, investiguer le service concerné et maintenir la revue humaine prévue en §7.2 ; ne pas transformer une prédiction en décision automatique. En cas de drift d'entrée ou de sortie, examiner les dossiers et collecter des annotations avant de conclure à une dégradation. Un candidat qui échoue aux garde-fous est rejeté et tracé ; une promotion acceptée produit un artefact séparé, à valider et déployer par une étape humaine.

**Limites et prérequis de mise en service :** alerting, ordonnanceur hebdomadaire, journalisation réelle des scores avec `request_id`, canal conseiller, encadrement de conservation des données et instrumentation OOD/calibration ne sont pas tous en place. La réserve métier bloquante de §7.3 et les conditions juridiques rappelées en §8 restent à lever ; le monitoring présenté ici ne vaut pas autorisation de mise en production.
